```python
from huggingface_hub import InferenceClient

client = InferenceClient(
    token=HF_token,
    model = "Qwen/Qwen2.5-7B-Instruct"
)

question = input("How may I help you? ")
dangerous_words = ["kill","suicide","end my life","to die","lethal dose","over dose"]

def is_validInput(question):
    for keyword in dangerous_words:
        if keyword in question:
            return False
    return True

if not is_validInput(question):
    print("I'm sorry i can't help with it please consult your doctor!")
else:
    messages = [
        {"role":"system","content":"Act like a helpful medical assistant"
        "make user friendly, clear ,safe and consice response"
        "dont give ambiguous and detailed response"
        "incase of any emergency throw a one line emergency response"
        "always recommend consulting a real doctor incase of severe issues"
        "never diagnose a disease or prescribe medications"
        "make responses clear and beneficial for users"},
        {"role": "user", "content": question}
    ]
            
    response = client.chat_completion(messages=messages, max_tokens=200)
    print(response.choices[0].message.content)

```
